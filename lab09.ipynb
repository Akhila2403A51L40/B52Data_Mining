```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(42)
X = np.linspace(0, 10, 50).reshape(-1, 1)
y = 2 + 3 * X - 0.5 * X**2 + np.random.normal(0, 1, X.shape)
```

```python
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
```

```python
model = LinearRegression()
model.fit(X_poly, y)
```

```python
y_pred = model.predict(X_poly)
```

```python
mse = mean_squared_error(y, y_pred)
print(f"Mean Squared Error: {mse}")
```

```python
plt.scatter(X, y, color='blue', label="Original Data")
plt.plot(X, y_pred, color='red', label="Polynomial Regression Line")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
```

```python
import numpy as np
import pandas as pd
np.random.seed(0)
X = np.random.rand(100, 1) * 10
y = 25000 + 5000 * X + np.random.randn(100, 1) * 2000
df = pd.DataFrame({'YearsExperience': X.flatten(), 'Salary': y.flatten()})
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['YearsExperience'], df['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Salary vs. Years of Experience')
plt.show()
```

Task 2

```python
X = df[['YearsExperience']]  # Independent variable
y = df['Salary']  # Dependent variable
```

```python
from sklearn.preprocessing import PolynomialFeatures

degree = 2  # Set the degree of the polynomial
poly_features = PolynomialFeatures(degree=degree)
X_poly = poly_features.fit_transform(X)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_poly, y)
```

Task 3

```python
y_pred = model.predict(X_poly)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

```python
plt.scatter(X, y, color='blue', label='Original Data')
plt.plot(X, y_pred, color='red', label='Polynomial Regression Curve')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Polynomial Regression')
plt.legend()
plt.show()
```
