```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Generate synthetic data
np.random.seed(42)  # Set seed for reproducibility
X = np.linspace(0, 10, 100)
y = 2 * X + 5 + np.random.normal(scale=1, size=100)  # Add noise

# --- Question 1: Impact of Noise ---
# Increase noise
y_noisy = 2 * X + 5 + np.random.normal(scale=3, size=100)
model_noisy = LinearRegression()
model_noisy.fit(X.reshape(-1, 1), y_noisy)
y_pred_noisy = model_noisy.predict(X.reshape(-1, 1))

# Plot with increased noise
plt.scatter(X, y_noisy, label='Data with Increased Noise')
plt.plot(X, y_pred_noisy, color='red', label='Regression Line (Noisy)')
plt.title('Regression with Increased Noise')
plt.legend()
plt.show()
# Observation: Increased noise makes the regression line less precise.
```
